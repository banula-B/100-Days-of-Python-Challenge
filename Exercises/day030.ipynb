# Exercise No: 02

A module is a single Python file (.py) that contains code such as functions, classes, or variables. For example, chess.py is a module.

A package is a folder that contains multiple Python modules and usually includes an ```__init__.py``` file. For example, games is a package containing chess.py and tic_tac_toe.py.

In simple words:
```text
Module = one Python file
Package = a folder containing related Python modules
```
