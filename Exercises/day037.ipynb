# Exercise No: 01

```python
grocery1 = ["Hammer", 2, 12.50]
grocery2 = ["Nails", 100, 0.10]
with open(r"..\Resources\csv files\inventory.csv", "a", newline="") as file:
    writer = csv.writer(file)

    
    writer.writerow(grocery1)
    writer.writerow(grocery2)
```

# Exercise No: 02

```python
total_value = 0
with open(r"..\Resources\csv files\inventory.csv", "r") as file:
    csv_reader = csv.DictReader(file)

    for row in csv_reader: 

        quantity = float(row["quantity"])
        price = float(row["price"])
        total_value += quantity * price

print(f"Total = {total_value}")
```
