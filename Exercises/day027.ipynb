# Exercise No: 01

```python
names = ["Alice", "Bob", "Anna", "Dave"]

name_list = list(filter(lambda name: name[0]=="A" , names))

print(name_list)
```

# Exercise No: 02

```python
from random import randint

# Create a number list with 10 random numbers between 1-30
numbers = [randint(1,100) for _ in range(10)]
print(numbers)

greater_than_fifty = list(filter(lambda x: x>50 , numbers))

print(greater_than_fifty)
```

# Exercise No: 03

```python
fruits = ["apple", "", "banana", "", "cherry"]

fruits_without_empty = list(filter(None, fruits))
print(fruits_without_empty)
```
