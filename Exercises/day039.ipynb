# Exercise No: 03

```python
number = random.randint(1,50)

print("Welcome to Number Guessing Game!")
print("I'm Thinking of a number between 1 and 50")
print("You have 5 attempts to guess the number")


for attempts in range(1,6):
    try:
        guess = int(input(f"Enter your Guess. You have{attempts}/5"))

        if guess == number:
            print("Your Guess is Correct. Congratulations!")
            break
        elif guess < number:
            print("Too Low!")
        else:
            print("Too High!")
    except ValueError:
        print("Please enter a valid number")

else:
    print("\nSorry, You used all 5 attempts")
    print(f"The correct answer is : {number}")
```

```python
s_number = random.randint(1,50)

print("Welcome to Number Guessing Game!")
print("I'm Thinking of a number between 1 and 50")
print("You have 5 attempts to guess the number")

attempts = 0


while attempts <5:
    try:
        guess = int(input("Enter Your Guess"))
        attempts +=1

        if guess == s_number:
            print("Your Guess is Correct")
            break

        elif guess < s_number:
            print("Too Low")

        else:
            print("Too high")
        print(f"your guessed number ={guess}")
        print(f"Remaining attempts ={5-attempts}")

    except ValueError:
        print("Please enter valid number")

else:
    print("Sorry, You have used all attempts.")
    print(f"Correct answer is {s_number}")
```
