This is for print 5*5 grid using nested loop

```python
# Print a 5 × 5 grid of hash symbols

for row in range(5):
    for col in range(5):
        print("#", end="")
    print()
```

# Exercise No: 02|

This is for print simple multiplication table using nested loop

```python
# Print a multiplication table from 1 to 5

for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end=" ")
    print()
```

# Exercise No: 03

This is for print combinations

```python
# Print every combination of categories and items

categories = ["Fruit", "Color"]
items = ["Apple", "Red"]

for category in categories:
    for item in items:
        print(category, "-", item)
```
