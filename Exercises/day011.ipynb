This exercise will print letter count in a given string list



```python
# Set the list

animals=["cat","dog","lion","elephant"]

sum = 0

#take string one by one from list
for animal in animals:
    # take letters one by one form animal
    for i in animal:
        #for each letter in animal, increase the value of sum by 2
        sum +=1


print(sum)
```
