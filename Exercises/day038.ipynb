# Exercise No: 02

```python
from time import strptime
from datetime import datetime

input_date = input("Enter Yor Birthday(DD/MM/YYYY): ")

birthday = datetime.strptime(input_date, "%d/%m/%Y")

print(f"You are born on :{birthday.strftime("%A")}")
```

# Exercise No: 03

```python
from datetime import datetime, timedelta

current_time = datetime.now()
future_time = current_time + timedelta(minutes=10000)

print("10,000 minutes from now:", future_time)
```
