# Exercise No: 01

This exercise write numbers from 10 to 1 using range() 

```python
for i in range(10,0,-1):
    print(i)
```

# Exercise No: 03

This return the sum of all numbers from 1 to whatever number you enter

```python
num = int(input("Enter your number: "))
sum=0
for i in range(1,num+1):
    sum+=i
print(sum)
```
