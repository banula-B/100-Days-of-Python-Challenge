```python
if not os.path.exists(r"../Resources/text files/counter.txt"):
    with open(r"../Resources/text files/counter.txt","w") as file:
        file.write("1\n")
else:
    with open(r"../Resources/text files/counter.txt","r") as file:
        line = file.readlines()
        # print(line)

        last_no = int(line[-1])
        # last_number = int(line[-1])
        new_no = last_no + 1
    with open(r"../Resources/text files/counter.txt","a") as file:
        file.write(f"{new_no}\n")
```
