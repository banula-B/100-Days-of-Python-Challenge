# Exercise No: 01

```python
class NegativeTemperatureError(Exception):
    pass

def check_temp(celsius):
    if celsius< -273.15:
        raise NegativeTemperatureError(
            "Temperature is Negative. Please enter valid temperature"
        )
    else:
        return "Tepurature is valid"



try:
    celsius = float(input("Enter Temperature: "))
    print(check_temp(celsius))

except NegativeTemperatureError as error:
    print(f"Error: {error}")

```

# Exercise No: 02

```python
try:
    num = int(input("Enter number multiple of 10: "))

    if num % 10 != 0:
        raise ValueError("Invalid number. please enter a number which multiple of 10!")
    else:
        print("Valid number.")

except ValueError as error:
    print(f"Error: {error}")
```

# Exercise No: 03

```python
class EmptyInputError(Exception):
    pass

while True:
    try:
        text = input("Enter Something: ")

        if not text.strip():
            raise EmptyInputError("You entered nothing. Please enter something!")
        else:
            print("Input is valid.")
        
            break

    except EmptyInputError as error:
        print(f"Error: {error}")
```
