# Exercise No: 01

```python
class NegativeTemperatureError(Exception):
    pass

def check_temp(celsius):
    if celsius <0:
        raise NegativeTemperatureError(
            "Enter a Tempurature abouve zero! "
        )
    return "Temperature is valid."

celsius = float(input("Enter tempurature in Celsius:"))

try:
    print(check_temp(celsius))
except NegativeTemperatureError as error:
    print(f"Error: {error}")


```

# Exercise No: 02

```python
try:
    num = int(input("Enter a number: "))

    if num%10 !=0:
        raise ValueError("The number must be a multiple of 10!.")
    else:
        print(f"{num} is a valid number")

except ValueError as error:
    print(f"Error: {error}")
```
