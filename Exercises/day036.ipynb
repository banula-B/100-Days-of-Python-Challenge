```python
import json
```

# Exercise No: 01

```python

book = {
    "title": "The Hobbit",
    "author": "J.R.R. Tolkien",
    "genres": ["Fantasy", "Adventure"],
    "available": True
}

json_string = json.dumps(book, indent = 2)

print(json_string)
```

# Exercise No: 02

```python
json_string = '{"item": "Laptop", "price": 999.99, "in_stock": true}'

python_string = json.loads(json_string)

discount_price = python_string["price"] * 0.9

python_string["price"] = discount_price

print(python_string)
```

# Exercise No: 03

```python
try:
    with open("high_scores.json", "r") as file:
        high_scores = json.load(file)

except FileNotFoundError:
    high_scores = {"Player1": 100}

    with open("high_scores.json", "w") as file:
        json.dump(high_scores, file, indent=4)

print(high_scores)
```
