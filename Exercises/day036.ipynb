```python
import json
```

# Exercise No: 01

```python

book = {
    "title": "The Hobbit",
    "author": "J.R.R. Tolkien",
    "genres": ["Fantasy", "Adventure"],
    "available": True
}

json_book = json.dumps(book, indent=2)

print(json_book)
```

# Exercise No: 02

```python
json_string = '{"item": "Laptop", "price": 999.99, "in_stock": true}'

product = json.loads(json_string)

discounted_price = int(product["price"]) * 0.90

product["price"] = discounted_price

print(product)
```

# Exercise No: 03

```python
try:
    with open("high_scores.json", "r") as file:
        high_scores = json.load(file)

except FileNotFoundError:
    high_scores = {"Player1": 100}

    with open("high_scores.json", "w") as file:
        json.dump(high_scores, file, indent=4)

print(high_scores)
```
