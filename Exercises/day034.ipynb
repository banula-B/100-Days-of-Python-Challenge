# Exercise No: 02

```python
list_of_item = ["Alfa Romeo","Bentley","Chevrolet"]
try:
    index = int(input("Enter index to print item: "))
    print(list_of_item[index])
except ValueError:
    print("Please enter a numerical value.")
except IndexError:
    print("Invalid index. Enter 0-2 value")
```
