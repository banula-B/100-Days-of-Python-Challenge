# Exercise No: 01

```python
try:
    age = int(input("Enter Your Age: "))
    print(f"{age}")
except ValueError:
    print("Please enter a numerical value.")
```

# Exercise No: 02

```python
list_of_item = ["Alfa Romeo","Bentley","Chevrolet"]
try:
    index = int(input("Enter index to print item: "))
    print(list_of_item[index])
except ValueError:
    print("Please enter a numerical value.")
except IndexError:
    print("Invalid index. Enter 0-2 value")
```

# Exercise No: 03

```python
def safe_divide(a, b):
    return a/b

try:
    num1 = int(input("Enter 1st Number: "))
    num2 = int(input("Enter 2nd Number: "))
    print(safe_divide(num1,num2))

except ValueError:
    print("Enter integers to do division!")

except ZeroDivisionError:
    print("Divide by Zero is not valid.\nPlease enter non zero value for 2nd Number")
```
