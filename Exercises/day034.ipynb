# Exercise No: 01

```python
try:
    age = int(input("Enter Your Age: "))
except ValueError:
    print("Please enter a numerical value.")

finally:
    print(f"Your age: {age}")
```

# Exercise No: 02

```python
list_of_item =["Banana","Apple","Orange"]

try:
    num = int(input("Enter index(1/2/3): "))
    print(list_of_item [num-1])
except ValueError:
    print("Please enter a numerical value.")
except IndexError:
    print("Your number outside the valid range")
```

```python
def safe_divide(a,b):
    return a/b

try:
    a = int(input("Enter value A: "))
    b = int(input("Enter value B: "))


    print(safe_divide(a,b))

except ZeroDivisionError:
    print("Can't divide ny zero")
    print("None")
```
