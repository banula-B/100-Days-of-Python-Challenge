```python
# Temperature checker using if, elif, else

# Get temperature in celsius from user
temperature = int(input("Enter the temperature in Celsius: "))


if temperature > 30:
    print("Hot")
elif temperature >= 20:
    print("Warm")
else:
    print("Cold")
```

```python
# Age checker using if, elif, else

# Get age from user
age = int(input("Enter your age: "))


if age < 13:
    print("Child")
elif age >= 13 and age <= 19:
    print("Teen")
else:
    print("Adult")
```

```python
# Simple number checker

# Get number from user
number = int(input("Enter a number: "))

# Check number and print category
if number < 10:
    print("Small")
elif number >= 10 and number <= 50:
    print("Medium")
else:
    print("Large")
```
