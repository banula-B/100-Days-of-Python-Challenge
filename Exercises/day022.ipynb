# Exercise No: 02

```python
def make_shirt(size, message):
    print(f"The shirt is size {size} and has the message '{message}' printed on it.")
```

```python
# Call by positional arguments

make_shirt("Large","WILD BEAR")
```

```python
# Call by keyword arguments

make_shirt(message="Hog Rider", size="Extra Large")
```
