```python
from functools import reduce
```

# Exercise No: 01

```python
numbers = [10, 25, 7, 42, 18]

maximum = reduce(lambda x, y: x if x > y else y, numbers)

print(maximum)
```

# Exercise No: 02

```python
from functools import reduce

words = ["Python", "is", "awesome"]

sentence = reduce(lambda x, y: x + " " + y, words)

print(sentence)
```

# Exercise No: 03

```python
total = reduce(lambda x, y: x + y, range(1, 11))

print(total)
```
