### Exercise No: 01

This is a program to find if a number between **1 and 100** (inclusive)

```python
a = int(input("Enter your number"))

if a >= 0 and a <=100:
    print("The number is between 1 and 100.")
else:
    print("The number is not-inclusive")
```
